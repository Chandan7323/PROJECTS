```python
import tkinter as tk
from time import strftime

# Create the root window
root = tk.Tk()
root.title("Digital Clock")

# Define the clock display style using the 7-segment concept (using font)
label = tk.Label(root, font=('Courier', 80), background='black', foreground='cyan')
label.pack(anchor='center')

# Function to update the time
def update_time():
    # Get the current time in HH:MM:SS format
    current_time = strftime('%H:%M:%S')
    # Update the label with the current time
    label.config(text=current_time)
    # Update the time every 1 second (1000 ms)
    label.after(1000, update_time)

# Call the function to update the time
update_time()

# Run the main loop of the Tkinter window
root.mainloop()
```
